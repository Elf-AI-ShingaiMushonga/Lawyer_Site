# Lawyer User Guide

This guide reflects the current DM-Inc lawyer site as of March 16, 2026.

## 1. Sign In and Access

1. Open the site root and click `Open Lawyer Site` on the landing page.
2. Sign in with your firm email address and password.
3. Internal MFA is temporarily disabled for the lawyer site, so no authenticator code is required.
4. After login you should arrive on `Dashboard`.

If login succeeds but a page or matter is blocked, your access may be limited by role, matter membership, or an ethical wall.

## 2. Main Areas Lawyers Should Know

- `Dashboard` for the priority inbox, task focus, and recent matters.
- `Matters` for matter search, filtering, and matter workspace access.
- `Calendar` for date-driven work and deadline ownership.
- `Billing` for invoice and payment visibility.
- `Time Entries` and `Timers` for fee capture.
- `Portal`, `Helpdesk`, `Knowledge Base`, and `Analytics` via the jump or command menu.

## 3. Recommended Daily Workflow

1. Start on `Dashboard` and review the `Priority Inbox` and `My Tasks`.
2. Open `Calendar` to check overdue and near-term deadlines.
3. Enter the correct file through `Matters` before doing substantive work.
4. Use the matter workspace to update notes, tasks, documents, stage, and next actions.
5. Capture time during the day in `Time Entries` or `Timers`.
6. Review billing context before end of day where relevant.
7. If an IT issue blocks work, open a `Helpdesk` ticket rather than sending informal email.

## 4. Matter Directory and Workspace

Use `Matters` to search by matter number, client, or title.

Inside a matter, lawyers should keep these areas current:

- matter summary, status, and stage,
- parties and contacts,
- notes and timeline entries,
- deadlines and calendar context,
- linked documents and workbench items,
- open tasks and closure items.

Use pinned and recently viewed matters to return to active files quickly.

## 5. Calendar and Deadlines

Use `Calendar` views (`My`, `Team`, `Matter`) to manage obligations.

1. Check overdue and due-soon work first.
2. Acknowledge deadlines once ownership is clear.
3. Use override actions only with a documented reason.
4. Keep the matter workspace and the calendar aligned when dates move.

## 6. Tasks and Approvals

Task work is embedded in matter execution and workflow management.

1. Create tasks directly from matters or from templates.
2. Assign multiple assignees only when work is genuinely collaborative.
3. Use dependencies and checklists to preserve sequence and quality control.
4. Use approvals before marking review-gated work complete.

## 7. Documents and DMS Workflows

Use matter-linked document tools for controlled drafting and retrieval.

1. Upload documents into the correct matter.
2. Keep version history intact instead of replacing files outside the system.
3. Lock documents during active drafting when collaboration needs tighter control.
4. Set document state and metadata correctly (`draft`, `reviewed`, `final`, `filed`, and related fields).
5. Use search and OCR-driven retrieval to find content quickly.
6. Use production and Bates tools only for formal disclosure or production workflows.

## 8. Time Capture, Expenses, and Billing

### Time Capture

1. Use timers for active work where practical.
2. Use `Time Entries` for manual capture and cleanup.
3. Add clear narratives and correct coding so entries are invoice-ready.
4. Submit entries for review and lock them when required by process.

### Expenses

1. Capture disbursements with the correct receipt.
2. Mark reimbursable status correctly.
3. Route expense items for approval before invoicing.

### Billing

1. Review invoice status, adjustments, and payment activity by matter.
2. Export invoice artifacts (PDF or LEDES) only when needed.
3. Track pending versus settled payments in the billing views.

## 9. Trust Accounting (Where Authorized)

Trust accounting is role-controlled and should be used only by authorised staff.

1. Post deposits, disbursements, and transfers to the correct ledger.
2. Do not bypass reconciliation workflows.
3. Review trust reports before period close.
4. Use reversal or correction flows rather than editing historical ledger entries directly.

## 10. CRM, Intake, and Conflict Checks

Use `CRM` and intake workflows when opening new work or reviewing prospects.

1. Create intake records with all known parties and entities.
2. Run conflict checks before proceeding.
3. Review both direct matches and semantic or OCR-assisted evidence.
4. Treat semantic hits as review-required evidence, not automatic disqualification.
5. Record override reasons clearly if work proceeds after a flagged conflict.

Note: semantic conflict scans and some AI-assisted operations run asynchronously. If a scan stays queued, ask operations to check the worker and scheduler services.

## 11. Portal, Knowledge Base, and Helpdesk

### Client Portal

1. Use the portal for controlled client messaging, uploads, invoices, and payments.
2. Share only approved client-facing matter artifacts.
3. Review portal uploads promptly and move relevant files into matter document flows.

### Knowledge Base

Use the knowledge base for internal process guidance, practice notes, and reusable operational knowledge.

### Helpdesk

1. Use `Helpdesk` for laptop, account, software, printer, dock, monitor, or access issues.
2. Link the affected asset where available.
3. Track ticket status in the queue instead of relying on informal follow-up.

## 12. Security and Compliance Rules

1. Never share login credentials or active sessions.
2. Internal MFA is currently disabled for the lawyer site, so password hygiene and logout discipline matter even more.
3. Do not access matters outside your assignment scope.
4. Respect legal holds, retention controls, and audit visibility.
5. Use in-app exports only when required and auditable.

## 13. Common Issues

### I cannot access a matter

Possible causes:
- you are not on the matter team,
- an ethical wall denies access,
- your current role does not permit that surface.

Action: contact your administrator to verify matter membership and access policy.

### Login works but a page is blocked

Action: confirm whether the page is role-restricted or matter-scoped before assuming the system is broken.

### Conflict scan remains queued

Action: ask operations to check worker and scheduler service health.

### I cannot upload a document

Possible causes:
- file size exceeds policy,
- file type is blocked,
- matter access is denied,
- the upload target is incorrect.

Action: reduce file size if needed, confirm file type, confirm matter access, and retry.

### I have an IT issue that blocks work

Action: open a `Helpdesk` ticket and include the affected asset if visible.

## 14. End-of-Day Checklist

Use this before sign-off:

1. Critical deadlines are acknowledged, escalated, or resolved.
2. Time entries are complete, descriptive, and tied to the correct matters.
3. Key documents are versioned and staged correctly.
4. New conflict checks and intake issues are reviewed with evidence.
5. Billing, portal, or helpdesk items requiring action are not left informal or untracked.
